```python
import pandas as pd

data = pd.read_json("outputs/vanilla-340M-4096-model-truncate_length_sweep.jsonl", lines=True)
softpick_data = pd.read_json("outputs/softpick-340M-4096-model-truncate_length_sweep.jsonl", lines=True)
data
```

```python
def expand_hidden_activation(data: pd.DataFrame) -> pd.DataFrame:
    out = data.copy()
    out["kurtosis"] = out["hidden_activation"].map(lambda x: x.get("kurtosis"))
    out["max_activation"] = out["hidden_activation"].map(lambda x: x.get("max"))
    out["min_activation"] = out["hidden_activation"].map(lambda x: x.get("min"))
    return out

data = expand_hidden_activation(data)
softpick_data = expand_hidden_activation(softpick_data)
data

```

```python
def expand_sink_rate(data: pd.DataFrame) -> pd.DataFrame:
    out = data.copy()
    out["sink_rate_0.2"] = out["sink_rate"].map(lambda x: x.get("0.2"))
    out["sink_rate_0.3"] = out["sink_rate"].map(lambda x: x.get("0.3"))
    return out

data = expand_sink_rate(data)
softpick_data = expand_sink_rate(softpick_data)
data["sink_rate_0.2"]
```

```python
print(data[["sequence_length", "sink_rate_0.3"]])
```

```python
import matplotlib.pyplot as plt

labels = {
    "x": "sequence_length",
    "y": "sink_rate_0.3",
    "title": "Sink rate (0.3) vs sequence length",
    "vanilla": "vanilla",
    "softpick": "softpick",
}

plot_data = data.sort_values("sequence_length")
plot_softpick = softpick_data.sort_values("sequence_length")
fig, ax = plt.subplots(figsize=(10, 4))

x_ticks = sorted(set(plot_data["sequence_length"]).union(set(plot_softpick["sequence_length"])))
x_positions = {value: index for index, value in enumerate(x_ticks)}

ax.plot(plot_data["sequence_length"].map(x_positions), plot_data["sink_rate_0.3"], marker="o", label=labels["vanilla"])
ax.plot(plot_softpick["sequence_length"].map(x_positions), plot_softpick["sink_rate_0.3"], marker="o", label=labels["softpick"])
ax.set_xlabel(labels["x"])
ax.set_ylabel(labels["y"])
ax.set_title(labels["title"])
ax.set_ylim(-0.05, 1.0)
ax.set_xticks(range(len(x_ticks)), x_ticks)
ax.legend()

```
